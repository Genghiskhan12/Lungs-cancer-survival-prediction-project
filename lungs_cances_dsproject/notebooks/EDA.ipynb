```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
df = pd.read_csv('../data/lung_cancer_dataset.csv')
df.head(10)
```

```python
df.info()
display(df.describe(include='all').round(2))
```

```python
missing = df.isnull().sum()
print(f'Total missing values: {missing.sum()}')
if missing.sum():
    display(missing[missing > 0].to_frame('Missing values'))
else:
    print('No missing values.')
```

```python
# Target variable distribution
target_counts = df['Survived'].value_counts().sort_index()
target_percentages = df['Survived'].value_counts(normalize=True).sort_index() * 100
display(pd.DataFrame({'Count': target_counts, 'Percentage': target_percentages.round(2)}))
sns.countplot(data=df, x='Survived', hue='Survived', palette='Set2', legend=False)
plt.title('Distribution of Survival Outcome')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()
```

```python
# Example categorical feature
counts = df['Gender'].value_counts()
plt.pie(counts, labels=counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('Set2'))
plt.title('Distribution of Gender')
plt.axis('equal')
plt.show()
```

```python
# Univariate analysis: numeric feature distributions
numeric_features = df.select_dtypes(include=np.number).columns.tolist()
display(df[numeric_features].describe().T.round(2))
ncols = 4
nrows = int(np.ceil(len(numeric_features) / ncols))
df[numeric_features].hist(bins=20, figsize=(16, max(12, nrows * 3)), layout=(nrows, ncols), edgecolor='black')
plt.suptitle('Distributions of Numeric Features', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Bivariate analysis: numeric feature relationships
numeric_df = df.select_dtypes(include=np.number)
correlation_matrix = numeric_df.corr()
display(correlation_matrix.round(2))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```

```python
# Numeric features most associated with survival
target_encoded = df['Survived'].map({'Yes': 1, 'No': 0})
correlation_df = numeric_df.drop(columns=['Survived'], errors='ignore').copy()
correlation_df['Survived'] = target_encoded
target_correlations = correlation_df.corr()['Survived'].drop('Survived').sort_values(key=abs, ascending=False)
display(target_correlations.to_frame('Correlation').round(3))
```
